### Usage example

First, need to instantiate the image classifier.

```python
from image_classification import ImageClassifier

ic = ImageClassifier(model_path="saved_models/EfficientNetV2S_100epochs_0p8dropout_dataset2_classweighted_best_val_loss.h5",
                     class_names=['Battery', 'Camera', 'Keyboard', 'Laptop', 'Microwave', 'Mobile', 'Mouse', 'PCB', 'Player', 'Printer', 'Smartwatch', 'Television', 'Washing Machine'])
```

Then proceed. Example usage:

```python
import matplotlib.pyplot as plt

img_path = "test_img_examples/galaxy_z_fold_7.webp"
img = plt.imread(img_path)
plt.imshow(img)
plt.axis('off')
plt.show()

pred, confidence = ic.predict_class(img_path)
print(pred, confidence)
```

NB: first prediction will be slow, subsequent predictions much quicker
